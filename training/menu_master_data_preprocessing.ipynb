```python

import pandas as pd
from pathlib import Path

# 원본 CSV 불러오기
dataset_dir = Path("../dataset")
input_path = dataset_dir / "menugen_menu_master.csv"

df = pd.read_csv(input_path, encoding="utf-8-sig", dtype={"fd_Code": "string"})

# 불필요한 번호 컬럼 삭제
df = df.drop(columns=["no"])

# 컬럼명 변경
df = df.rename(columns={
    "fd_Code": "menuCode",
    "upper_Fd_Grupp_Nm": "mainCategory",
    "fd_Grupp_Nm": "subCategory",
    "fd_Nm": "menuName",
    "fd_Wgh": "weight",
    "food_Cnt": "foodCount"
})

# 문자열 컬럼의 앞뒤 공백 제거
text_columns = ["menuCode", "mainCategory", "subCategory", "menuName"]

for column in text_columns:
    df[column] = df[column].str.strip()

# 사용할 컬럼 순서 정리
df = df[
    ["menuCode", "mainCategory", "subCategory",
     "menuName", "weight", "foodCount"]
]

# 전처리 결과 확인
print("데이터 크기:", df.shape)
print(df.dtypes)
display(df.head())

# 전처리한 CSV 저장
output_path = dataset_dir / "MENU_MASTER_v1.csv"
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print('저장 완료!')
```
