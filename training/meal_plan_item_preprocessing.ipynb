```python
import pandas as pd

# 1. CSV 파일 읽기
file_path = 'D:\\HUMAN-meal-planner-ai\\backend\\src\\main\\java\\com\\human\\backend\\cost\\dummy\\meal_plan_item.csv'
df = pd.read_csv(file_path)

print(df.head())  # 데이터프레임의 처음 5행을 출력하여 확인
```

```python



# 2. display_order에 따른 식판 칸 위치(slot_type) 부여 함수
def assign_slot_type(row):
    order = row['display_order']
    if order == 1:
        return 'RICE'       # 1번 칸: 밥
    elif order == 2:
        return 'SOUP'       # 2번 칸: 국/찌개
    elif order == 3:
        return 'MAIN'  # 3번 칸: 메인 반찬
    else:
        return 'SIDE'   # 4, 5번 칸 등: 서브 반찬

# slot_type 컬럼 생성
df['slot_type'] = df.apply(assign_slot_type, axis=1)

# 3. 데이터베이스 테이블 컬럼 순서에 맞게 재정렬
cols = ['plan_item_id', 'plan_id', 'slot_type', 'menu_id', 'display_order', 'created_at', 'updated_at']
df = df[cols]

df.loc[df['menu_id'] == 1, 'menu_id'] = 2  # 1번을 2번으로 변경

# 4. 파일 저장 (덮어쓰기)
df.to_csv(file_path, index=False)

print("식판 칸 순서에 맞게 slot_type이 성공적으로 수정되었습니다!")
```
